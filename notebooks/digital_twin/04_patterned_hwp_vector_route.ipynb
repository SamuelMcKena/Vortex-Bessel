# Patterned-HWP Vector Route

Continuous, six-wedge, and tiled patterned half-wave-plate routes compared at the axicon input.

STAGE88: editable controls

```python
from pathlib import Path
import matplotlib
matplotlib.use('Agg', force=True)
import matplotlib.pyplot as plt
import vbb_study as pkg
from vbb_study.digital_twin.nathan_vector_hexagon import NathanHexagonConfig, PatternedHWPConfig, run_patterned_hwp_route

assert Path(pkg.__file__).is_absolute()
publication = Path(pkg.__file__).resolve().parents[1]
fig_dir = publication / 'outputs' / 'figures' / 'digital_twin'
fig_dir.mkdir(parents=True, exist_ok=True)
cfg = NathanHexagonConfig.fast(grid_n=128, z_planes=5)
```

```python
routes = [
    run_patterned_hwp_route(cfg, hwp=PatternedHWPConfig(case='continuous'), route_id='continuous'),
    run_patterned_hwp_route(cfg, hwp=PatternedHWPConfig(case='six_wedges'), route_id='six_wedges'),
    run_patterned_hwp_route(cfg, hwp=PatternedHWPConfig(case='mosaic', tiles_per_sector=8), route_id='mosaic_8'),
]
assert routes[0].comparison.normalized_rms_error < 1e-12
assert routes[1].comparison.normalized_rms_error > routes[2].comparison.normalized_rms_error
for route in routes:
    print(route.route_id, route.comparison.complex_overlap, route.comparison.normalized_rms_error)
```

```python
labels = [r.route_id for r in routes]
errors = [r.comparison.normalized_rms_error for r in routes]
fig, ax = plt.subplots(figsize=(6, 3.5), constrained_layout=True)
ax.bar(labels, errors, color=['#2a9d8f', '#e76f51', '#457b9d'])
ax.set_ylabel('normalised RMS field error')
ax.set_title('Patterned-HWP route comparison at axicon input')
fig.savefig(fig_dir / 'nathan_vector_hexagon_fig02_patterned_hwp_routes.png', dpi=200)
plt.close(fig)
print(fig_dir / 'nathan_vector_hexagon_fig02_patterned_hwp_routes.png')
```
