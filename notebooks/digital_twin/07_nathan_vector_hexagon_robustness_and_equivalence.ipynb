```python
from pathlib import Path
import csv
import matplotlib
matplotlib.use('Agg', force=True)
import matplotlib.pyplot as plt
import vbb_study as pkg
from vbb_study.digital_twin.nathan_vector_hexagon import NathanHexagonConfig, hwp_robustness_sweep, serial_slm_robustness_sweep, build_route_comparison_report, route_conclusion

assert Path(pkg.__file__).is_absolute()
publication = Path(pkg.__file__).resolve().parents[1]
fig_dir = publication / 'outputs' / 'figures' / 'digital_twin'
csv_dir = publication / 'outputs' / 'csv' / 'digital_twin'
fig_dir.mkdir(parents=True, exist_ok=True)
csv_dir.mkdir(parents=True, exist_ok=True)
cfg = NathanHexagonConfig.fast(grid_n=96, z_planes=5)
```

```python
hwp_rows = hwp_robustness_sweep(cfg, family='tiles_per_sector', values=(1, 2, 4, 8, 16))
slm_rows = serial_slm_robustness_sweep(cfg, family='fill_factor', values=(1.0, 0.93, 0.85))
assert hwp_rows[0]['normalized_rms_error'] > hwp_rows[-1]['normalized_rms_error']
for name, rows in [('nathan_vector_hexagon_hwp_tile_sweep.csv', hwp_rows), ('nathan_vector_hexagon_slm_fill_factor_sweep.csv', slm_rows)]:
    path = csv_dir / name
    with path.open('w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=list(rows[0].keys()))
        writer.writeheader(); writer.writerows(rows)
    print(path)
report = build_route_comparison_report(cfg)
print(route_conclusion(report))
```

```python
fig, ax = plt.subplots(figsize=(6, 3.5), constrained_layout=True)
ax.plot([row['value'] for row in hwp_rows], [row['normalized_rms_error'] for row in hwp_rows], '-o', label='HWP tile error')
ax.set_xlabel('tiles per 60 degree sector')
ax.set_ylabel('normalised RMS field error')
ax.set_title('Patterned-HWP mosaic convergence')
ax.legend()
fig.savefig(fig_dir / 'nathan_vector_hexagon_fig05_robustness_sweeps.png', dpi=200)
plt.close(fig)
print(fig_dir / 'nathan_vector_hexagon_fig05_robustness_sweeps.png')
```
