# 09 - Nathan MODE 1: ideal P2 field through the inherited downstream Digital Twin

**MODE 1** injects the *ideal* canonical Nathan six-sector `VectorField` at the common **P2** handoff plane and
propagates it through the **inherited** downstream Digital Twin geometry (vector axicon -> current F0 scalar
per-component focus bridge -> vector ASM sample z-stack), with **F2** (vectorial pupil-spectrum) as a diagnostic.

MODE 1 **bypasses** patterned HWP, SLM1, relay, intermediate HWP, SLM2, final QWP, panel realism, carrier/iris,
HWP mosaics and waveplate errors. It therefore **only partially represents the physical bench** (downstream only;
upstream physical generation bypassed). It tests whether the source-validated (V0, docs/53) sixfold / dark-core
hexagonal Bessel field **survives** the inherited micro-scale optics. It is **not** a physical HWP/QWP/SLM
realisation (that is MODE 2A/2B).

Metrics deliberately **separate** three things: a **C3-vs-C6 symmetry classification** (with a triangular
veto), *source-like dark-core survival*, and *clean single-wall usefulness*. Order-6 ring content **alone**
does not imply a hexagon (a C3/triangular field carries a strong order-6 harmonic on a ring). A machine-
readable `simulation_scope_manifest.json` is written for the run.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from vbb_study.digital_twin import (
    NathanHexagonConfig, run_mode1_ideal_p2_downstream,
    plot_mode1_p2_input_diagnostics, plot_mode1_sample_region,
    plot_mode1_f0_vs_f2, plot_mode1_v0_to_mode1_scale,
)

OUT = Path('..') / '..' / 'outputs' / 'figures' / 'digital_twin' / 'nathan_mode1_ideal_p2'
cfg = NathanHexagonConfig.fast(grid_n=192, z_planes=15, angular_samples=720)
mode1 = run_mode1_ideal_p2_downstream(cfg, run_f2=True, f2_chunk_size=128, output_dir=OUT)
print('MODE 1 manifest ->', OUT / 'simulation_scope_manifest.json')
print('represents_physical_bench:', mode1.manifest['represents_physical_bench'])
print('suggested outcome:', mode1.completion['suggested_outcome'])
```

## MODE 1 outcome

**M1-B** - the ideal Nathan six-sector P2 field **does not** produce a visually convincing micro-scale
**hexagonal** beam under the inherited downstream Digital Twin geometry. The output is dominated by a
**dark-core triangular / three-lobed (C3) structure**: order-6 ring content is present (and even exceeds
order-3), but the 120-degree rotational self-similarity exceeds the 60-degree one and the dark core is
visibly triangular, so it fails the hexagonal veto. **Physical HWP/QWP/SLM realisation remains paused.**

Notes:
- a *structured dark-core field does survive* (hence M1-B, not M1-D) - it is the wrong symmetry class, not an
  absence of structure;
- the field is closer to hexagonal in the *early* formation planes and **degrades to triangular** downstream -
  this is the lead for the next task (an ideal geometry/parameter sensitivity study);
- centre-treatment is robust (the dark core is real, not a P2 grid artefact) and F0/F2 agree.

This notebook **does not** run HWP / QWP / SLM generation, panel realism, carrier/iris, robustness sweeps or route
ranking. MODE 2A/2B remain **blocked**; MODE 3 out of scope.
