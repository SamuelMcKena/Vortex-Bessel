# Nathan MODE 1B - Geometry Target Search

This notebook calls package functions only. MODE 1B compares ideal downstream geometry candidates against the validated V0 Nathan hexagonal Bessel total-intensity target, while keeping visual success, feasibility, and current-lab realism separate.

```python
from pathlib import Path
from pprint import pprint

from vbb_study.digital_twin import (
    audit_mode1b_ring_count_planes,
    build_mode1b_target_template,
    run_mode1b_geometry_search,
    write_mode1b_ring_count_audit,
    write_mode1b_geometry_search_outputs,
)

OUTPUT_DIR = Path("outputs/figures/digital_twin/nathan_mode1b_geometry_search")
```

```python
audit = audit_mode1b_ring_count_planes()
audit_paths = write_mode1b_ring_count_audit(audit, OUTPUT_DIR)
pprint(audit.conclusion)
pprint(audit_paths)
audit.rows
```
