# Shared Axicon Hexagon Propagation

All routes feed the same vector axicon and air-side propagation model.

STAGE88: editable controls

```python
from pathlib import Path
import matplotlib
matplotlib.use('Agg', force=True)
import matplotlib.pyplot as plt
import vbb_study as pkg
from vbb_study.digital_twin.nathan_vector_hexagon import NathanHexagonConfig, build_route_comparison_report, write_report_csv

assert Path(pkg.__file__).is_absolute()
publication = Path(pkg.__file__).resolve().parents[1]
fig_dir = publication / 'outputs' / 'figures' / 'digital_twin'
csv_dir = publication / 'outputs' / 'csv' / 'digital_twin'
fig_dir.mkdir(parents=True, exist_ok=True)
csv_dir.mkdir(parents=True, exist_ok=True)
cfg = NathanHexagonConfig.fast(grid_n=96, z_planes=5)
```

```python
report = build_route_comparison_report(cfg)
paths = write_report_csv(report, csv_dir)
assert len(report.output_rows) == 6
for key, path in paths.items():
    print(key, path)
for row in report.output_rows:
    print(row['route_id'], row['best_h6'], row['best_wall_continuity'], row['best_core_darkness'])
```

```python
fig, ax = plt.subplots(figsize=(7, 3.5), constrained_layout=True)
ax.bar([row['route_id'] for row in report.output_rows], [row['best_h6'] for row in report.output_rows], color='#457b9d')
ax.set_ylabel('best H6')
ax.tick_params(axis='x', rotation=25)
ax.set_title('Shared axicon output: route-by-route H6')
fig.savefig(fig_dir / 'nathan_vector_hexagon_fig04_shared_axicon_h6.png', dpi=200)
plt.close(fig)
print(fig_dir / 'nathan_vector_hexagon_fig04_shared_axicon_h6.png')
```
