# Serial Dual-SLM Vector Route

Ideal and panel-realistic serial SLM routes compared against the same canonical target.

STAGE88: editable controls

```python
from pathlib import Path
import matplotlib
matplotlib.use('Agg', force=True)
import matplotlib.pyplot as plt
import vbb_study as pkg
from vbb_study.digital_twin.nathan_vector_hexagon import NathanHexagonConfig, run_serial_slm_route

assert Path(pkg.__file__).is_absolute()
publication = Path(pkg.__file__).resolve().parents[1]
fig_dir = publication / 'outputs' / 'figures' / 'digital_twin'
fig_dir.mkdir(parents=True, exist_ok=True)
cfg = NathanHexagonConfig.fast(grid_n=128, z_planes=5)
```

```python
ideal = run_serial_slm_route(cfg, case='ideal', route_id='ideal_serial_slm')
realistic = run_serial_slm_route(cfg, case='panel_realistic', route_id='panel_realistic_serial_slm')
wrong = run_serial_slm_route(cfg, case='ideal', naive_psi2=True, route_id='wrong_phase_serial_slm')
assert ideal.comparison.normalized_rms_error < 1e-12
assert wrong.comparison.normalized_rms_error > 0.1
assert realistic.metadata['order_filter_applied'] is True
for route in [ideal, realistic, wrong]:
    print(route.route_id, route.comparison.complex_overlap, route.comparison.normalized_rms_error)
```

```python
routes = [ideal, realistic, wrong]
fig, ax = plt.subplots(figsize=(7, 3.5), constrained_layout=True)
ax.bar([r.route_id for r in routes], [r.comparison.complex_overlap for r in routes], color=['#2a9d8f', '#e9c46a', '#e76f51'])
ax.set_ylabel('complex-vector overlap')
ax.set_ylim(0, 1.05)
ax.tick_params(axis='x', rotation=15)
ax.set_title('Serial dual-SLM route target overlap')
fig.savefig(fig_dir / 'nathan_vector_hexagon_fig03_serial_slm_routes.png', dpi=200)
plt.close(fig)
print(fig_dir / 'nathan_vector_hexagon_fig03_serial_slm_routes.png')
```
