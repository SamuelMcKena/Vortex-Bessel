# Nathan Vector Hexagon Target Field

Canonical six-sector radial/azimuthal target at the shared axicon-input plane.

STAGE88: editable controls

```python
from pathlib import Path
import matplotlib
matplotlib.use('Agg', force=True)
import matplotlib.pyplot as plt
import numpy as np
import vbb_study as pkg
from vbb_study.digital_twin.nathan_vector_hexagon import NathanHexagonConfig, canonical_target_diagnostics

assert Path(pkg.__file__).is_absolute()
publication = Path(pkg.__file__).resolve().parents[1]
fig_dir = publication / 'outputs' / 'figures' / 'digital_twin'
fig_dir.mkdir(parents=True, exist_ok=True)
GRID_N = 128
cfg = NathanHexagonConfig.fast(grid_n=GRID_N, z_planes=5)
diag = canonical_target_diagnostics(cfg)
field = diag['field']
assert field.ex.shape == (GRID_N, GRID_N)
assert set(np.unique(diag['sector_mask'])) == {0, 1}
print(pkg.__file__)
print('target_power', field.power)
```

```python
extent = [field.grid['x'][0] / 1e-3, field.grid['x'][-1] / 1e-3, field.grid['x'][0] / 1e-3, field.grid['x'][-1] / 1e-3]
plus = diag['circular_plus']
panels = [
    (diag['sector_mask'], 'sector mask', 'tab20'),
    (diag['polarization_angle_rad'], 'polarisation angle rad', 'twilight'),
    (diag['stokes']['S0'], 'S0', 'viridis'),
    (np.abs(plus), '|E plus|', 'magma'),
]
fig, axes = plt.subplots(2, 2, figsize=(8, 7), constrained_layout=True)
for ax, (arr, title, cmap) in zip(axes.ravel(), panels):
    im = ax.imshow(arr, origin='lower', extent=extent, cmap=cmap)
    ax.set_title(title)
    ax.set_xlabel('x mm')
    ax.set_ylabel('y mm')
    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
fig.savefig(fig_dir / 'nathan_vector_hexagon_fig01_target_field.png', dpi=200)
plt.close(fig)
print(fig_dir / 'nathan_vector_hexagon_fig01_target_field.png')
```
